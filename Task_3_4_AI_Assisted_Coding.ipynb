Build a small image classification application using the trained CIFAR-10 CNN model.

The application will allow the user to upload an image and receive a predicted class.

```python
from google.colab import files

uploaded = files.upload()
```

```python
import gradio as gr
import tensorflow as tf
import numpy as np

# Load the trained Keras model
model = tf.keras.models.load_model("cifar10_cnn_model.keras")

# CIFAR-10 class names
class_names = [
    "airplane",
    "automobile",
    "bird",
    "cat",
    "deer",
    "dog",
    "frog",
    "horse",
    "ship",
    "truck"
]

def predict_image(image):
    # Resize image to 32x32
    image = image.resize((32, 32))

    # Ensure RGB format
    image = image.convert("RGB")

    # Convert image to NumPy array
    image = np.array(image)

    # Normalize pixel values
    image = image.astype("float32") / 255.0

    # Add batch dimension
    image = np.expand_dims(image, axis=0)

    # Make prediction
    predictions = model.predict(image, verbose=0)

    # Get predicted class
    predicted_class = class_names[np.argmax(predictions[0])]

    return predicted_class


# Create Gradio interface
demo = gr.Interface(
    fn=predict_image,
    inputs=gr.Image(type="pil", label="Upload an Image"),
    outputs=gr.Textbox(label="Predicted CIFAR-10 Class"),
    title="CIFAR-10 Image Classifier",
    description="Upload an image and the CNN model will predict its CIFAR-10 class."
)

# Launch the application
demo.launch()
```
