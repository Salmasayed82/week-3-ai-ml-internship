## 1. Reasoning Prompt

### Initial Prompt

A student studies for 3 hours every day for 5 days.
How many total hours does the student study?

## 1. Reasoning Prompt

### Initial Prompt

A student studies for 3 hours every day for 5 days.
How many total hours does the student study?

### Initial Result

3 hours × 5 days = 15 hours. ✅

Answer: 15 hours

### Evaluation

The prompt produced a correct and clear answer.
However, it is a simple reasoning problem, so the prompt does not explicitly ask the model to explain the reasoning step by step.

### Improved Prompt

A student studies for 3 hours every day for 5 days.
Solve the problem step by step and explain the calculation clearly.
Then provide the final answer.

### Improved Result

Step 1: Identify the information
The student studies 3 hours each day.
The student studies for 5 days.
Step 2: Multiply

To find the total number of hours, multiply the hours per day by the number of days:

3 × 5 = 15

Step 3: State the answer

The student studies for 15 hours in total. ✅

Final Answer: 15 hours

### Comparison

The initial prompt produced the correct answer, but it gave only a short explanation.

The improved prompt asked the model to solve the problem step by step. As a result, the response was more structured and clearly showed the calculation before giving the final answer.

The improved prompt was more effective for a reasoning task because it encouraged a step-by-step explanation.

### Initial Result

    return sum(numbers) / len(numbers)
numbers = [10, 20, 30]

20.0
### Evaluation

The prompt generated a Python function that addresses the requested task.
However, it does not specify how the function should handle an empty list.

### Initial Result

The generated function was tested with:

```python
numbers = [10, 20, 30]
print(calculate_average(numbers))

### Improved Result

```python
def calculate_average(numbers):
    if not numbers:
        return 0

    return sum(numbers) / len(numbers)


# Example
numbers = [10, 20, 30]
average = calculate_average(numbers)

print(average)
